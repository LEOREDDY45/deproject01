```python
constructors_schema=StructType(fields=[
    StructField('constructorId', IntegerType(), False),
    StructField('constructorRef', StringType(), True),
    StructField('name', StringType(), True),
    StructField('nationality', StringType(), True),
    StructField('url', StringType(), True)
])

   
```

```python
constructors=spark.read.option('header', True).schema(constructors_schema).json(bronze+'/constructors.json')
```

```python
constructors=constructors.withColumnRenamed("constructorId", "constructor_id")\
                         .withColumnRenamed("constructorRef", "constructor_ref")\
                         .withColumn('ingestion_date', current_date()).drop('url')
```

```python
constructors.write.mode('overwrite').parquet(silver+'/constructors')
```
