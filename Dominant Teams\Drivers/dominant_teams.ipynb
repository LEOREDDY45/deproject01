```python
df=spark.read.parquet(gold + '/race_results_calculated')
```

```python
df=df.select('race_year', 
             'position', 
             'points', 
             'team',
             'calculated_points')\
            .withColumn('decade', floor(col('race_year').cast('int')/10)) 
```

```python
df=   df.groupBy('team','decade')\
        .agg(count('*').alias('total_races'), 
             sum("calculated_points").alias("total_points"), 
             round(avg("calculated_points"),2).alias("avg_points"))\
        .filter(col("total_races") >= 50)\
        .withColumn('decade', col('decade')*10)
```

```python
df.write.mode('overwrite').parquet(gold+'/dominant_teams_decade')
```
