```python
laptimes_schema= StructType(fields=[
    StructField("raceId", IntegerType(), False),
    StructField("driverId", IntegerType(), False),
    StructField("position", IntegerType(), True),
    StructField("lap", IntegerType(), True),
    StructField("time", StringType(), True),
    StructField("milliseconds", IntegerType(), True)
])
```

```python
laptimes=spark.read.schema(laptimes_schema).option('header', True).csv(bronze+'/laptimes')
```

```python
laptimes=laptimes.withColumnRenamed('raceId','race_id')\
                 .withColumnRenamed('driverId','driver_id')\
                 .withColumn('ingestion_date',current_timestamp())
```

```python
laptimes.write.mode('overwrite').parquet(silver+'/lap_times')
```
