```python
cricuits_schema=StructType(fields=[
    StructField('circuitId', IntegerType(), False),
    StructField('circuitRef', StringType(), True),
    StructField('name', StringType(), True),
    StructField('location', StringType(), True),
    StructField('country', StringType(), True),
    StructField('lat', FloatType(), True),
    StructField('lng', FloatType(), True),
    StructField('alt', IntegerType(), True),
    StructField('url', StringType(), True)
])
```

```python
circuits=spark.read.schema(cricuits_schema).option('header', True).csv(bronze+"/circuits.csv")
```

```python
circuits=circuits.withColumnRenamed('circuitId', 'circuit_id')\
                 .withColumnRenamed('circuitRef', 'circuit_ref')\
                 .withColumnRenamed('lat', 'latitude')\
                 .withColumnRenamed('lng', 'longitude')\
                 .withColumnRenamed('alt', 'altitude')\
                 .withColumn('ingestion_date', current_date())
```

```python
circuits=circuits.drop('url')
```

```python
circuits.write.mode('overwrite').parquet(silver+'/circuits_tb')
```
